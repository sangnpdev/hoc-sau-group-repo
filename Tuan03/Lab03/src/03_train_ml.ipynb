Đánh giá Ridge và Gradient Boosting bằng cross-validation thật.


## Chuẩn bị


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

OUTPUT = PROJECT_ROOT / "outputs"
OUTPUT.mkdir(exist_ok=True)

```

## Hàm: `log`


```python
def log(message):
    print(message)
    with (OUTPUT / "train_log.txt").open("a", encoding="utf-8") as file:
        file.write(f"{message}\n")

```

## Chạy notebook


```python
import pandas as pd
from IPython import get_ipython
get_ipython().run_line_magic("run", str(PROJECT_ROOT / "house_price_pipeline.ipynb"))

log("[4] Training Ridge và GradientBoosting...")
train, _ = load_competition_data(PROJECT_ROOT / "data")
specs = [
    ExperimentSpec("engineered_ridge", "engineered", "ridge"),
    ExperimentSpec("engineered_gradient_boosting", "engineered", "gradient_boosting"),
]
results = []
for spec in specs:
    result = cross_validate_pipeline(train, spec, n_splits=5, quick=True)
    log(f"{spec.model_name}: CV log-RMSE={result['log_rmse_mean']:.6f}")
    results.append(result)

pd.DataFrame(results).sort_values("log_rmse_mean").to_csv(OUTPUT / "ml_results.csv", index=False)
```
