## Chuẩn bị


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

OUTPUT = PROJECT_ROOT / "outputs"
OUTPUT.mkdir(exist_ok=True)

```

## Hàm: `log`


```python
def log(message):
    print(message)
    with (OUTPUT / "train_log.txt").open("a", encoding="utf-8") as file:
        file.write(f"{message}\n")

```

## Chạy notebook


```python
import numpy as np
from IPython import get_ipython
get_ipython().run_line_magic("run", str(PROJECT_ROOT / "house_price_pipeline.ipynb"))

log("[3] Preprocessing...")
train, _ = load_competition_data(PROJECT_ROOT / "data")
features = make_feature_frame(train, "engineered")
numeric = features.select_dtypes(include=[np.number, "bool"]).columns.tolist()
categorical = [column for column in features.columns if column not in numeric]
transformed = make_preprocessor(features).fit_transform(features)
log(f"Numerical features: {len(numeric)}")
log(f"Categorical features: {len(categorical)}")
log(f"Transformed shape: {transformed.shape}")
```
