## 2. Chạy hai biến thể

Thiết lập mặc định giữ source: basic chạy 15 epoch; utilities chạy tối đa 100 epoch và early stopping patience 5. Kết quả lưu trong `/content/fer2013_pytorch_results`.

```python
!python fer2013_emotion_cnn.py \
  --variant both \
  --train-csv fer2013_training_onehot.csv \
  --validation-csv fer2013_publictest_onehot.csv \
  --output-dir /content/fer2013_pytorch_results \
  --device auto
```

## 3. Xem kết quả và learning curve

```python
import json
from pathlib import Path
from IPython.display import display, Image

results_dir = Path('/content/fer2013_pytorch_results')
summary = json.loads((results_dir / 'summary.json').read_text())
for variant, metrics in summary.items():
    print(variant, {k: metrics[k] for k in ('epochs_completed', 'final_publictest_loss', 'final_publictest_accuracy', 'training_time_minutes')})
    display(Image(filename=str(results_dir / variant / 'training_history.png')))
    display(Image(filename=str(results_dir / variant / 'sample_prediction.png')))
```
